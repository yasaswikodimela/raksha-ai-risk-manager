```python
import pandas as pd
import numpy as np

print("RAKSHA - IEEE-CIS Fraud Detection")
print("----------------------------------")

transaction_df = pd.read_csv(
    "../data/raw/train_transaction.csv"
)

print("Dataset loaded successfully!")
print("Shape:", transaction_df.shape)
```

```python
print("Fraud Distribution")
print("===================")

print(transaction_df["isFraud"].value_counts())

print("\nFraud Percentage")
print("=================")

print(
    transaction_df["isFraud"]
    .value_counts(normalize=True) * 100
)
```

```python
identity_df = pd.read_csv(
    "../data/raw/train_identity.csv"
)

print("Identity dataset loaded successfully!")
print("Shape:", identity_df.shape)
```

```python
df = transaction_df.merge(
    identity_df,
    on="TransactionID",
    how="left"
)

print("Merged dataset successfully!")
print("Shape:", df.shape)
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print("Top 20 columns with missing values:")
print(missing.head(20))
```

```python
df["TransactionDay"] = (
    df["TransactionDT"] // (24 * 60 * 60)
).astype("int32")

df["TransactionHour"] = (
    (df["TransactionDT"] // (60 * 60)) % 24
).astype("int8")

print("Time features created successfully!")

print(
    df[
        [
            "TransactionDT",
            "TransactionDay",
            "TransactionHour"
        ]
    ].head()
)
```

```python
df["AmountLog"] = np.log1p(df["TransactionAmt"])

df["AmountDecimal"] = (
    (df["TransactionAmt"] - np.floor(df["TransactionAmt"])) * 100
).round(2)

print("Amount features created successfully!")

print(
    df[
        [
            "TransactionAmt",
            "AmountLog",
            "AmountDecimal"
        ]
    ].head()
)
```

```python
df = df.sort_values("TransactionDT").reset_index(drop=True)

n = len(df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_df = df.iloc[:train_end].copy()
val_df = df.iloc[train_end:val_end].copy()
test_df = df.iloc[val_end:].copy()

print("Train shape:", train_df.shape)
print("Validation shape:", val_df.shape)
print("Test shape:", test_df.shape)

print("\nFraud rates:")
print("Train:", train_df["isFraud"].mean())
print("Validation:", val_df["isFraud"].mean())
print("Test:", test_df["isFraud"].mean())
```

```python
feature_cols = [
    "TransactionAmt",
    "TransactionDay",
    "TransactionHour",

    "ProductCD",

    "card1",
    "card2",
    "card3",
    "card4",
    "card5",
    "card6",

    "addr1",
    "addr2",

    "dist1",
    "dist2",

    "P_emaildomain",
    "R_emaildomain",

    "C1",
    "C2",
    "C4",
    "C5",
    "C6",
    "C7",
    "C8",
    "C9",
    "C10",

    "D1",
    "D2",
    "D3",
    "D4",
    "D5",

    "V1",
    "V2",
    "V3",
    "V4",
    "V5",
    "V6",
    "V7",
    "V8",
    "V9",
    "V10",

    "M1",
    "M2",
    "M3",
    "M4",
    "M5",
    "M6",
    "M7",
    "M8",
    "M9",

    "DeviceType",
    "DeviceInfo"
]

print("Number of selected features:", len(feature_cols))
```

```python
X_train = train_df[feature_cols].copy()
X_val = val_df[feature_cols].copy()
X_test = test_df[feature_cols].copy()

y_train = train_df["isFraud"].copy()
y_val = val_df["isFraud"].copy()
y_test = test_df["isFraud"].copy()

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)

print("\ny_train:", y_train.shape)
print("y_val:", y_val.shape)
print("y_test:", y_test.shape)
```

```python
categorical_cols = [
    "ProductCD",
    "card1",
    "card2",
    "card3",
    "card4",
    "card5",
    "card6",
    "addr1",
    "addr2",
    "P_emaildomain",
    "R_emaildomain",
    "DeviceType",
    "DeviceInfo"
]

for col in categorical_cols:
    freq = X_train[col].value_counts(dropna=False)

    X_train[col] = X_train[col].map(freq).fillna(0)
    X_val[col] = X_val[col].map(freq).fillna(0)
    X_test[col] = X_test[col].map(freq).fillna(0)

print("Categorical features encoded successfully!")
print("X_train shape:", X_train.shape)
```

```python
print("Data types before training:")
print(X_train.dtypes.value_counts())
```

```python
string_cols = X_train.select_dtypes(include=["object", "string"]).columns

print("Remaining string columns:")
print(list(string_cols))

for col in string_cols:
    X_train[col] = pd.to_numeric(X_train[col], errors="coerce")
    X_val[col] = pd.to_numeric(X_val[col], errors="coerce")
    X_test[col] = pd.to_numeric(X_test[col], errors="coerce")

print("\nString columns converted!")
print(X_train.dtypes.value_counts())
```

```python
print("Missing values in X_train:")
print(X_train.isna().sum().sort_values(ascending=False).head(15))

print("\nTotal missing values:")
print(X_train.isna().sum().sum())
```

```python
print("Infinite values in X_train:", np.isinf(X_train.select_dtypes(include=np.number)).sum().sum())
print("Infinite values in X_val:", np.isinf(X_val.select_dtypes(include=np.number)).sum().sum())
print("Infinite values in X_test:", np.isinf(X_test.select_dtypes(include=np.number)).sum().sum())
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

model = HistGradientBoostingClassifier(
    learning_rate=0.08,
    max_iter=200,
    max_leaf_nodes=31,
    min_samples_leaf=50,
    l2_regularization=1.0,
    class_weight="balanced",
    early_stopping=False,
    random_state=42
)

print("Training RAKSHA fraud detection model...")

model.fit(
    X_train,
    y_train
)

print("Training completed successfully!")
```

```python
# Remove features that contain no useful information
valid_features = [
    col for col in X_train.columns
    if X_train[col].notna().sum() > 0
    and X_train[col].nunique(dropna=True) > 1
]

print("Original features:", X_train.shape[1])
print("Usable features:", len(valid_features))

X_train = X_train[valid_features]
X_val = X_val[valid_features]
X_test = X_test[valid_features]

print("New X_train shape:", X_train.shape)
```

```python
print("Generating validation predictions...")

y_val_prob = model.predict_proba(X_val)[:, 1]
y_val_pred = (y_val_prob >= 0.5).astype(int)

print("Predictions generated successfully!")
print("Number of validation transactions:", len(y_val_pred))
print("Predicted fraud transactions:", y_val_pred.sum())
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("RAKSHA - Validation Performance")
print("--------------------------------")

print(classification_report(
    y_val,
    y_val_pred,
    target_names=["Legitimate", "Fraud"],
    digits=4
))

cm = confusion_matrix(y_val, y_val_pred)

print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.10, 0.20, 0.30, 0.40, 0.50]

print("RAKSHA - Threshold Analysis")
print("---------------------------")

for threshold in thresholds:
    pred = (y_val_prob >= threshold).astype(int)

    precision = precision_score(y_val, pred)
    recall = recall_score(y_val, pred)
    f1 = f1_score(y_val, pred)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1: {f1:.4f}"
    )
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(0.05, 0.96, 0.05)

results = []

for threshold in thresholds:
    pred = (y_val_prob >= threshold).astype(int)

    results.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(y_val, pred),
        "Recall": recall_score(y_val, pred),
        "F1": f1_score(y_val, pred)
    })

threshold_df = pd.DataFrame(results)

print("RAKSHA - Detailed Threshold Analysis")
print("------------------------------------")
print(threshold_df.to_string(index=False))

best_row = threshold_df.loc[threshold_df["F1"].idxmax()]

print("\nBest threshold by F1:")
print(best_row)
```

```python
from sklearn.metrics import confusion_matrix

print("RAKSHA - False Positive Analysis")
print("--------------------------------")

thresholds = np.arange(0.50, 0.96, 0.05)

for threshold in thresholds:
    pred = (y_val_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        pred
    ).ravel()

    print(
        f"Threshold: {threshold:.2f} | "
        f"FP: {fp:5d} | "
        f"FN: {fn:4d} | "
        f"TP: {tp:4d} | "
        f"TN: {tn:5d}"
    )
```

```python
print("RAKSHA - Cost Sensitive Threshold Analysis")
print("------------------------------------------")

FP_COST = 10
FN_COST = 100

thresholds = np.arange(0.50, 0.96, 0.05)

for threshold in thresholds:
    pred = (y_val_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        pred
    ).ravel()

    total_cost = (fp * FP_COST) + (fn * FN_COST)

    print(
        f"Threshold: {threshold:.2f} | "
        f"FP: {fp:5d} | "
        f"FN: {fn:4d} | "
        f"Cost: ₹{total_cost:,}"
    )
```

```python
selected_threshold = 0.65

y_val_risk_pred = (y_val_prob >= selected_threshold).astype(int)

print("RAKSHA - Selected Threshold Performance")
print("---------------------------------------")
print("Selected threshold:", selected_threshold)

print("\nClassification Report:")
print(classification_report(
    y_val,
    y_val_risk_pred,
    target_names=["Legitimate", "Fraud"],
    digits=4
))

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_risk_pred))
```

```python
def assign_risk_level(probability):
    if probability < 0.30:
        return "LOW"
    elif probability < 0.65:
        return "MEDIUM"
    else:
        return "HIGH"


risk_levels = pd.Series(
    y_val_prob
).apply(assign_risk_level)

print("RAKSHA - Risk Level Distribution")
print("--------------------------------")

print(risk_levels.value_counts())

print("\nRisk Level Percentages:")
print(
    (risk_levels.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
risk_analysis = pd.DataFrame({
    "RiskLevel": risk_levels,
    "ActualFraud": y_val.values
})

risk_summary = risk_analysis.groupby("RiskLevel").agg(
    Transactions=("ActualFraud", "count"),
    FraudCases=("ActualFraud", "sum"),
    FraudRate=("ActualFraud", "mean")
)

risk_summary["FraudRate"] = (
    risk_summary["FraudRate"] * 100
).round(2)

risk_summary = risk_summary.loc[
    ["LOW", "MEDIUM", "HIGH"]
]

print("RAKSHA - Risk Level Fraud Analysis")
print("----------------------------------")
print(risk_summary)
```

```python
total_fraud = y_val.sum()

risk_summary["FraudCapture"] = (
    risk_summary["FraudCases"] / total_fraud * 100
).round(2)

print("RAKSHA - Fraud Capture by Risk Level")
print("-------------------------------------")
print(risk_summary[[
    "Transactions",
    "FraudCases",
    "FraudRate",
    "FraudCapture"
]])
```

```python
print("Generating final TEST predictions...")

y_test_prob = model.predict_proba(X_test)[:, 1]

print("TEST predictions generated successfully!")
print("Number of test transactions:", len(y_test_prob))
```

```python
selected_threshold = 0.65

y_test_pred = (y_test_prob >= selected_threshold).astype(int)

print("RAKSHA - FINAL TEST PERFORMANCE")
print("--------------------------------")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=["Legitimate", "Fraud"],
    digits=4
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python
def assign_risk_level(probability):
    if probability < 0.30:
        return "LOW"
    elif probability < 0.65:
        return "MEDIUM"
    else:
        return "HIGH"


test_risk_levels = pd.Series(
    y_test_prob
).apply(assign_risk_level)

test_risk_analysis = pd.DataFrame({
    "RiskLevel": test_risk_levels,
    "ActualFraud": y_test.values
})

test_risk_summary = test_risk_analysis.groupby("RiskLevel").agg(
    Transactions=("ActualFraud", "count"),
    FraudCases=("ActualFraud", "sum"),
    FraudRate=("ActualFraud", "mean")
)

test_risk_summary["FraudRate"] = (
    test_risk_summary["FraudRate"] * 100
).round(2)

test_risk_summary = test_risk_summary.loc[
    ["LOW", "MEDIUM", "HIGH"]
]

print("RAKSHA - FINAL TEST RISK ANALYSIS")
print("---------------------------------")
print(test_risk_summary)
```

```python
total_test_fraud = y_test.sum()

test_risk_summary["FraudCapture"] = (
    test_risk_summary["FraudCases"] / total_test_fraud * 100
).round(2)

print("RAKSHA - FINAL TEST FRAUD CAPTURE")
print("---------------------------------")
print(test_risk_summary[[
    "Transactions",
    "FraudCases",
    "FraudRate",
    "FraudCapture"
]])
```
